How Do People Read Dashboards?

Data Analysis Notebook | Dashboard Interpretation & Decision-Making Study

This notebook contains the exploratory data analysis (EDA) of survey responses collected to understand how users interpret a Logistics Operations Performance Dashboard.

The analysis explores:
- Dashboard interpretation
- Elements that attracted participants' attention first
- Sections participants spent the most time examining
- Operational concerns identified
- Actions participants would prioritize
- Qualitative feedback and thematic analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Dashboard Interpretation Study 2.csv')
```

```python
# Count responses by profession
profession_counts = df["What is your profession or area of expertise? (e.g Data Analyst)"].value_counts(dropna=False)

print(profession_counts)

```

```python
attention_counts = df["What caught your attention first? "].value_counts()

plt.figure(figsize=(9,5))

attention_counts.plot(kind="barh")

# Put the highest count at the top
plt.gca().invert_yaxis()

# Show whole numbers on the x-axis
plt.xticks(range(0, attention_counts.max() + 1))

plt.title("What Caught Participants' Attention First?")
plt.xlabel("Number of Participants")
plt.ylabel("Dashboard Element")
plt.savefig("attention_first.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
dashboard_counts = df["Without reading the title, what do you think this dashboard is about? "].value_counts()

plt.figure(figsize=(9,5))

dashboard_counts.plot(kind="barh")

plt.gca().invert_yaxis()

plt.title("Participants' Interpretation of the Dashboard")
plt.xlabel("Number of Participants")
plt.ylabel("Dashboard Interpretation")
plt.savefig("dashboard_purpose.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
section_counts = df["Which section did you spend the most time looking at?"].value_counts()

plt.figure(figsize=(9,5))

section_counts.plot(kind="barh")

# Put the highest count at the top
plt.gca().invert_yaxis()

# Display whole numbers on the x-axis
plt.xticks(range(0, section_counts.max() + 1))

plt.title("Dashboard Sections Participants Spent the Most Time Viewing")
plt.xlabel("Number of Participants")
plt.ylabel("Dashboard Section")

# Save image
plt.savefig("section_viewed_longest.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
concern_counts = df["If you were managing this operations, what would concern you most?"].value_counts()

plt.figure(figsize=(8,5))

concern_counts.sort_values().plot(kind="barh")

plt.title("Operational Concerns Identified by Participants")
plt.xlabel("Number of Participants")
plt.ylabel("Concern")
plt.savefig("operational_concern.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
action_counts = df["What action would you take first? "].value_counts()

plt.figure(figsize=(8,5))

action_counts.sort_values().plot(kind="barh")

plt.title("First Action Participants Would Take")
plt.xlabel("Number of Participants")
plt.ylabel("Action")
plt.savefig("first_action.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
confidence_counts = df["How confident are you that you understand what this dashboard is communicating? "].value_counts().sort_index()

plt.figure(figsize=(8,5))
confidence_counts.plot(kind='bar')

plt.title("Confidence in Interpreting the Dashboard")
plt.xlabel("Confidence Level")
plt.ylabel("Number of Participants")

plt.tight_layout()
plt.show()
```
